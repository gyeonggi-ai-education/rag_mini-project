```python
import zipfile, re, html

def load_hwpx(path):
    # 1. HWPX(zip)에서 본문 XML 읽기
    with zipfile.ZipFile(path) as z:
        xml = z.read("Contents/section0.xml").decode("utf-8")

    # 2. 문단(<hp:p>) 단위로 자르기
    paras = re.findall(r"<hp:p .*?</hp:p>", xml, flags=re.S)

    # 3. 각 문단에서 글자(<hp:t>)만 이어 붙이기
    result = []
    for p in paras:
        text = "".join(re.findall(r"<hp:t>(.*?)</hp:t>", p, flags=re.S))
        text = html.unescape(text).rstrip()   # &lt; → <
        if text.strip():                      # 빈 문단 제외
            result.append(text)
    return result
```

```python
paras = load_hwpx("../data/raw/ai_basic_act.hwpx")
print(len(paras))
for p in paras[:20]:
    print(repr(p))
```
